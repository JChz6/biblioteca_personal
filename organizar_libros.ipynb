```python

# Ejemplo de uso
if __name__ == "__main__":
    ruta = input("Introduce la ruta de tu biblioteca digital: ").strip()
    
    if os.path.isdir(ruta):
        df_libros = recopilar_libros(ruta)
        print(f"\nArchivos encontrados: {len(df_libros)}")
        print(df_libros.head())

        # Guardamos el resultado en CSV para trabajar después
        salida = os.path.join(ruta, "catalogo_biblioteca.csv")
        df_libros.to_csv(salida, index=False, encoding="utf-8-sig")
        print(f"\nCatálogo guardado en: {salida}")
    else:
        print("Ruta no válida.")
```
